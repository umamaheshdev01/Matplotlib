## 2. Series — 1D Data
> A **Series** is a one-dimensional labeled array — like a column in a spreadsheet.
> Every element has an **index** (label) and a **value**.

```python
# ── Creating a Series ──
s1 = pd.Series([10, 20, 30, 40, 50])               # Default index: 0,1,2,3,4
s2 = pd.Series([10, 20, 30], index=['a', 'b', 'c']) # Custom index
s3 = pd.Series({'Mon': 100, 'Tue': 200, 'Wed': 150}) # From a dictionary
s4 = pd.Series(99, index=['x', 'y', 'z'])            # Constant value

print("Default index:\n", s1)
print("\nCustom index:\n", s2)
print("\nFrom dict:\n", s3)
print(s4)
```

```python
# ── Accessing Elements ──
s = pd.Series([10, 20, 30, 40, 50], index=['a','b','c','d','e'])

print("By label  s['c']  :", s['c'])          # 30 — label access
print("By pos    s.iloc[2]:", s.iloc[2])      # 30 — position access
print("Slice     s['b':'d']:\n", s['b':'d'])  # label slice (inclusive!)
print("Multiple  s[['a','e']]:\n", s[['a','e']])

# ── Series Attributes ──
print("\nValues :", s.values)   # NumPy array of values
print("Index  :", s.index)     # Index labels
print("dtype  :", s.dtype)     # Data type
print("Name   :", s.name)      # Series name (None by default)
s.name = 'my_series'
print("Named  :", s.name)
```

```python
# ── Series Operations (vectorized — no loops needed) ──
s = pd.Series([10, 20, 30, 40, 50])

print("s * 2     :", (s * 2).values)
print("s + 100   :", (s + 100).values)
print("s > 25    :", (s > 25).values)   # Boolean Series
print("s[s > 25] :", s[s > 25].values)  # Filtered
print("Sum       :", s.sum())
print("Mean      :", s.mean())
print("Describe:\n", s.describe())       # Quick summary stats
```

## 6. Selecting & Indexing
> The most important skill in Pandas — selecting exactly the data you need.

```python
df = pd.read_csv('employees.csv')

# ── Selecting Columns ──
print("Single column (Series):\n", df['name'])             # Returns Series
print("\nMultiple columns (DataFrame):\n",
      df[['name', 'salary']])                               # Returns DataFrame
print("\nDot notation:", df.name.values)                   # Works for simple names
```

```python
# ── .loc  → Label-based selection (use row label & column name) ──
df_indexed = df.set_index('name')   # Set 'name' as the row index
print("=== .loc — label based ===")

print(df_indexed.loc['Alice'])                      # All cols for Alice
print(df_indexed.loc['Alice', 'salary'])             # Specific cell: 55000
print(df_indexed.loc[['Alice','Diana'], ['salary','score']])  # Rows & cols
print(df_indexed.loc['Alice':'Charlie', :])          # Slice (inclusive!)
```

```python
# ── .iloc → Integer position-based selection (like NumPy slicing) ──
df2 = pd.read_csv('employees.csv')
print("=== .iloc — position based ===")

print(df2.iloc[0])              # First row (all columns)
print(df2.iloc[0, 3])           # Row 0, Col 3  → salary value
print(df2.iloc[1:4])            # Rows 1,2,3 (stop is exclusive)
print(df2.iloc[:, 1:3])         # All rows, columns 1 and 2
print(df2.iloc[[0, 2, 4], :])   # Specific rows by position
```

```python
# ── Key rule: .loc is INCLUSIVE on both ends, .iloc is EXCLUSIVE on stop ──
s = pd.Series([10, 20, 30, 40, 50], index=['a','b','c','d','e'])

print(".loc ['a':'c'] (inclusive):", s.loc['a':'c'].values)   # [10 20 30]
print(".iloc [0:3]    (exclusive):", s.iloc[0:3].values)      # [10 20 30]
```

## 7. Filtering Data
> Extract rows that meet specific conditions — no loops needed.

```python
df = pd.read_csv('employees.csv')

# ── Single Condition ──
it_staff = df[df['dept'] == 'IT']
print("IT department:\n", it_staff)

high_earners = df[df['salary'] > 70000]
print("\nSalary > 70000:\n", high_earners)
```

```python
# ── Multiple Conditions ──
# Use & (AND), | (OR), ~ (NOT) — wrap each condition in parentheses!

it_high_score = df[(df['dept'] == 'IT') & (df['score'] > 85)]
print("IT dept AND score > 85:\n", it_high_score)

hr_or_finance = df[(df['dept'] == 'HR') | (df['dept'] == 'Finance')]
print("\nHR or Finance:\n", hr_or_finance)

not_it = df[~(df['dept'] == 'IT')]   # Everyone except IT
print("\nNot IT:\n", not_it)
```

```python
# ── Handy Filtering Methods ──

# .isin()  → match any value in a list
tech_finance = df[df['dept'].isin(['IT', 'Finance'])]
print(".isin(['IT','Finance']):\n", tech_finance)

# .between() → range filter (inclusive)
mid_salary = df[df['salary'].between(60000, 80000)]
print("\n.between(60k, 80k):\n", mid_salary)

# .str.contains()  → string matching
has_a = df[df['name'].str.contains('a', case=False)]  # Names containing 'a'
print("\nNames containing 'a':\n", has_a[['name', 'dept']])

# .query()  → SQL-like string syntax (clean & readable)
result = df.query("dept == 'IT' and salary > 70000")
print("\n.query() — IT and salary>70k:\n", result)
```

## 9. Handling Missing Data
> Real-world data is messy. This is one of the most important topics in Pandas.

```python
# Create a dataset with missing values
data = pd.DataFrame({
    'name'  : ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'age'   : [25, None, 35, 28, None, 40],
    'salary': [55000, 72000, None, 68000, 51000, None],
    'dept'  : ['HR', 'IT', 'IT', None, 'HR', 'Finance'],
    'score' : [88.5, 92.0, 78.3, 95.1, None, 76.4]
})
print("Dataset with missing values:\n", data)
```

```python
# ── Detecting Missing Values ──
print("=== isna() — True where missing ===")
print(data.isna())

print("\n=== Missing count per column ===")
print(data.isna().sum())

print("\n=== Missing % per column ===")
print((data.isna().mean() * 100).round(1))

print("\n=== Rows with ANY missing value ===")
print(data[data.isna().any(axis=1)])
```

```python
# ── Dropping Missing Values ──

# Drop rows where ANY column is NaN
dropped_any = data.dropna()
print("Drop rows with any NaN:", len(dropped_any), "rows left")

# Drop rows where ALL columns are NaN
dropped_all = data.dropna(how='all')
print("Drop rows with all NaN:", len(dropped_all), "rows left")

# Drop rows where specific columns have NaN
dropped_salary = data.dropna(subset=['salary', 'score'])
print("Drop where salary or score is NaN:\n", dropped_salary[['name','salary','score']])

# Drop columns with NaN
dropped_cols = data.dropna(axis=1)
print("\nDrop columns with NaN:", dropped_cols.columns.tolist())
```

```python
# ── Filling Missing Values ──

# Fill with a constant
filled_zero = data['salary'].fillna(0)
print("Fill with 0:    ", filled_zero.values)

# Fill with mean (most common for numeric)
salary_mean = data['salary'].mean()
filled_mean = data['salary'].fillna(salary_mean)
print(f"Fill with mean ({salary_mean:.0f}):", filled_mean.values)

# Fill with median (more robust to outliers)
filled_median = data['salary'].fillna(data['salary'].median())

# Fill with mode (best for categorical)
filled_mode = data['dept'].fillna(data['dept'].mode()[0])
print("Fill dept with mode:", filled_mode.values)

# Forward fill: use previous valid value (great for time-series)
filled_ffill = data['age'].ffill()
print("Forward fill age:   ", filled_ffill.values)

# Backward fill: use next valid value
filled_bfill = data['age'].bfill()
print("Backward fill age:  ", filled_bfill.values)
```

## 12. String Operations
> The `.str` accessor gives you vectorized string methods — no loops needed.

```python
names = pd.Series(['  Alice Smith  ', 'bob jones', 'CHARLIE BROWN', 'diana Prince'])

# ── Case ──
print("upper     :", names.str.upper().values)
print("lower     :", names.str.lower().values)
print("title     :", names.str.title().values)   # Capitalize each word
print("strip     :", names.str.strip().values)   # Remove leading/trailing spaces

# ── Inspect ──
print("\nlen       :", names.str.strip().str.len().values)
print("contains 'a':", names.str.contains('a', case=False).values)
print("startswith:", names.str.strip().str.startswith('A').values)

# ── Transform ──
print("\nreplace 'o'→'0':", names.str.replace('o', '0', regex=False).values)
print("split on ' '   :", names.str.strip().str.split(' ').values)

# ── Extract first name only ──
first_names = names.str.strip().str.split(' ').str[0].str.title()
print("First names:", first_names.values)
```

## 13. Applying Functions
> Transform data using custom logic — the bridge between Pandas and pure Python.

```python
df = pd.read_csv('employees.csv')

# ── map() → element-wise on a Series ──
dept_code = {'HR': 1, 'IT': 2, 'Finance': 3}
df['dept_code'] = df['dept'].map(dept_code)   # Map values using a dict
print("=== map() — dept codes ===")
print(df[['name', 'dept', 'dept_code']])
```

```python
# ── apply() → apply a function along rows or columns ──
df = pd.read_csv('employees.csv')

# Apply to a column (Series) — same as map for simple cases
df['salary_band'] = df['salary'].apply(
    lambda x: 'High' if x >= 80000 else ('Mid' if x >= 60000 else 'Low')
)
print("=== apply() on column ===")
print(df[['name', 'salary', 'salary_band']])

# Apply row-wise (axis=1) — access multiple columns in one function
def combined_score(row):
    """Custom scoring: mix performance score and seniority."""
    seniority_bonus = 5 if row['age'] >= 35 else 0
    return round(row['score'] + seniority_bonus, 1)

df['final_score'] = df.apply(combined_score, axis=1)
print("\n=== apply() row-wise ===")
print(df[['name', 'age', 'score', 'final_score']])
```

```python
# ── applymap() / map() on entire DataFrame (element-wise) ──
nums = pd.DataFrame({'A': [1.234, 5.678], 'B': [9.012, 3.456]})

# Round every element to 1 decimal
rounded = nums.map(lambda x: round(x, 1))   # pandas >= 2.1: .map(); older: .applymap()
print("=== map() on DataFrame ===")
print(rounded)

# ── np.where for fast conditional column ──
df = pd.read_csv('employees.csv')
df['veteran'] = np.where(df['age'] >= 35, 'Veteran', 'Junior')
print("\n=== np.where — fast conditional ===")
print(df[['name', 'age', 'veteran']])
```

## 14. GroupBy & Aggregation
> The most powerful tool for summarizing data by categories — think SQL's `GROUP BY`.

```python
df = pd.read_csv('employees.csv')

# ── Basic GroupBy ──
grouped = df.groupby('dept')    # Creates a GroupBy object

print("=== Mean salary per department ===")
print(grouped['salary'].mean().round(0))

print("\n=== Multiple stats on salary ===")
print(grouped['salary'].agg(['mean', 'min', 'max', 'count']).round(0))
```

```python
# ── agg() with different stats per column ──
summary = df.groupby('dept').agg(
    avg_salary  = ('salary', 'mean'),
    max_salary  = ('salary', 'max'),
    avg_score   = ('score',  'mean'),
    headcount   = ('name',   'count'),
    avg_age     = ('age',    'mean')
).round(1)

print("=== Department Summary ===")
print(summary)
```

```python
# ── Group by multiple columns ──
df['senior'] = df['age'] >= 30

multi_group = df.groupby(['dept', 'senior'])['salary'].mean().round(0)
print("=== Avg salary by dept and seniority ===")
print(multi_group)

# .unstack() → pivot the inner index level to columns
print("\n=== Unstacked (prettier) ===")
print(multi_group.unstack())
```

```python
# ── transform() → compute group stats but keep original row count ──
# (Unlike agg which collapses rows, transform returns same-size output)

df = pd.read_csv('employees.csv')

# Add dept average salary as a new column
df['dept_avg_salary'] = df.groupby('dept')['salary'].transform('mean').round(0)
df['vs_dept_avg']     = df['salary'] - df['dept_avg_salary']  # Above/below average

print("=== Individual vs Department Average ===")
print(df[['name','dept','salary','dept_avg_salary','vs_dept_avg']]
      .sort_values('dept'))
```

```python
# ── filter() → keep only groups that meet a condition ──

# Keep only departments with average salary > 65000
rich_depts = df.groupby('dept').filter(lambda g: g['salary'].mean() > 65000)
print("=== Depts with avg salary > 65k ===")
print(rich_depts[['name','dept','salary']])
```

## 15. Merging & Joining DataFrames
> Combining multiple tables — the Pandas equivalent of SQL JOINs.

```python
# Create two related tables
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4, 5],
    'name'  : ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'dept_id': [10, 20, 20, 30, 10]
})

departments = pd.DataFrame({
    'dept_id'  : [10, 20, 30, 40],
    'dept_name': ['HR', 'IT', 'Finance', 'Marketing'],
    'location' : ['NYC', 'SF', 'Chicago', 'LA']
})

print("Employees:\n", employees)
print("\nDepartments:\n", departments)
```

```python
# ── pd.merge() — the main join function ──

# INNER JOIN — only matching rows in both tables
inner = pd.merge(employees, departments, on='dept_id', how='inner')
print("=== INNER JOIN ===")
print(inner)

# LEFT JOIN — all employees, fill NaN if no dept match
left = pd.merge(employees, departments, on='dept_id', how='left')
print("\n=== LEFT JOIN ===")
print(left)

# RIGHT JOIN — all departments, NaN for employees with no dept
right = pd.merge(employees, departments, on='dept_id', how='right')
print("\n=== RIGHT JOIN ===")
print(right)

# OUTER JOIN — all rows from both tables
outer = pd.merge(employees, departments, on='dept_id', how='outer')
print("\n=== OUTER JOIN ===")
print(outer)
```

```python
# ── Merge on different column names ──
df_a = pd.DataFrame({'employee_id': [1,2,3], 'name': ['A','B','C']})
df_b = pd.DataFrame({'emp_id':      [1,2,4], 'score': [90,85,78]})

merged = pd.merge(df_a, df_b,
                   left_on='employee_id',   # Column in left DataFrame
                   right_on='emp_id',       # Column in right DataFrame
                   how='left')
print("=== Merge with different key names ===")
print(merged)

# ── Concatenation (stacking DataFrames) ──
df1 = pd.DataFrame({'name': ['Alice','Bob'],   'dept': ['HR','IT']})
df2 = pd.DataFrame({'name': ['Charlie','Diana'],'dept': ['IT','Finance']})

stacked = pd.concat([df1, df2], ignore_index=True)  # Stack rows
print("\n=== pd.concat() — stack rows ===")
print(stacked)

side_by_side = pd.concat([df1, df2], axis=1)   # Stack columns
print("\n=== pd.concat(axis=1) — stack columns ===")
print(side_by_side)
```

## 16. Reshaping — Pivot, Melt, Stack
> Transforming the shape of your table — wide ↔ long format.

```python
# Sample sales data in LONG format (each row = one observation)
sales = pd.DataFrame({
    'month'  : ['Jan','Jan','Jan','Feb','Feb','Feb','Mar','Mar','Mar'],
    'product': ['A','B','C','A','B','C','A','B','C'],
    'revenue': [120, 85, 95, 140, 90, 110, 130, 100, 125]
})
print("Long format:\n", sales)
```

```python
# ── pivot_table → Long → Wide (like Excel pivot tables) ──
pivot = sales.pivot_table(
    values='revenue',    # What to put in cells
    index='month',       # Rows
    columns='product',   # Columns
    aggfunc='sum'        # How to aggregate (mean, sum, count...)
)
print("=== pivot_table (Wide format) ===")
print(pivot)

# Add totals (margins)
pivot_with_totals = sales.pivot_table(
    values='revenue', index='month', columns='product',
    aggfunc='sum', margins=True, margins_name='Total'
)
print("\n=== With Totals ===")
print(pivot_with_totals)
```

```python
# ── melt → Wide → Long (the reverse of pivot) ──
wide = pivot.reset_index()   # Month becomes a column again
print("Wide format:\n", wide)

long = wide.melt(
    id_vars='month',              # Column(s) to keep as-is
    value_vars=['A','B','C'],     # Columns to unpivot
    var_name='product',           # Name for the variable column
    value_name='revenue'          # Name for the value column
)
print("\n=== melt (back to Long format) ===")
print(long.sort_values(['month','product']).reset_index(drop=True))
```

```python
# ── stack / unstack → pivot index levels ──
print("=== stack — move columns into index ===")
stacked = pivot.stack()   # Columns (A,B,C) become an inner index level
print(stacked.head(6))

print("\n=== unstack — move index level to columns ===")
unstacked = stacked.unstack()   # Reverse
print(unstacked)
```

```python
# ── Resampling — aggregate by time period ──
rng = np.random.default_rng(42)
dates  = pd.date_range('2024-01-01', periods=90, freq='D')
values = rng.normal(100, 15, 90).cumsum() + 1000  # Simulated daily revenue

ts = pd.Series(values, index=dates, name='revenue')

# Resample to weekly and monthly summaries
weekly  = ts.resample('W').mean().round(1)    # Weekly average
monthly = ts.resample('ME').sum().round(0)    # Monthly total

print("=== Daily → Weekly average ===")
print(weekly.head(4))
print("\n=== Daily → Monthly total ===")
print(monthly)

# ── Rolling window — moving average ──
ts_df = ts.to_frame()
ts_df['7d_avg']  = ts.rolling(window=7).mean().round(1)   # 7-day moving avg
ts_df['30d_avg'] = ts.rolling(window=30).mean().round(1)  # 30-day moving avg
print("\n=== Rolling averages ===")
print(ts_df.head(10))
```

## 18. Practical Mini-Projects

```python
# ════════════════════════════════════════════════════
# 🏪 Project 1: Sales Data Analysis Pipeline
# ════════════════════════════════════════════════════
rng = np.random.default_rng(7)

# Generate synthetic sales data
n = 200
products  = rng.choice(['Laptop','Phone','Tablet','Monitor','Keyboard'], n)
regions   = rng.choice(['North','South','East','West'], n)
qty       = rng.integers(1, 11, n)
unit_price = {'Laptop':1200,'Phone':800,'Tablet':500,'Monitor':400,'Keyboard':80}

sales_df = pd.DataFrame({
    'date'    : pd.date_range('2024-01-01', periods=n, freq='B'),  # Business days
    'product' : products,
    'region'  : regions,
    'qty'     : qty,
    'discount': rng.choice([0, 0.05, 0.10, 0.15], n),
})
sales_df['unit_price'] = sales_df['product'].map(unit_price)
sales_df['revenue']    = (sales_df['qty'] * sales_df['unit_price']
                           * (1 - sales_df['discount'])).round(2)
sales_df['month']      = sales_df['date'].dt.month_name()

print("=== Sample Sales Data ===")
print(sales_df.head())
print(f"\nTotal rows: {len(sales_df)} | Columns: {list(sales_df.columns)}")
```

```python
# ── Analysis 1: Revenue by Product ──
product_summary = sales_df.groupby('product').agg(
    total_revenue = ('revenue', 'sum'),
    total_qty     = ('qty', 'sum'),
    avg_discount  = ('discount', 'mean'),
    num_orders    = ('revenue', 'count')
).round(2).sort_values('total_revenue', ascending=False)

product_summary['revenue_share%'] = (
    product_summary['total_revenue'] / product_summary['total_revenue'].sum() * 100
).round(1)

print("=== Revenue by Product ===")
print(product_summary)
```

```python
# ── Analysis 2: Regional Performance ──
region_pivot = sales_df.pivot_table(
    values='revenue',
    index='region',
    columns='product',
    aggfunc='sum',
    margins=True,
    margins_name='TOTAL'
).round(0).fillna(0)

print("=== Revenue by Region & Product ===")
print(region_pivot)
```

```python
# ── Analysis 3: Monthly Trend ──
monthly_trend = (
    sales_df.set_index('date')
    .groupby(pd.Grouper(freq='ME'))['revenue']  # Group by month-end
    .agg(['sum', 'count', 'mean'])
    .round(2)
    .rename(columns={'sum': 'total_revenue', 'count': 'num_orders', 'mean': 'avg_order'})
)
monthly_trend['mom_growth%'] = monthly_trend['total_revenue'].pct_change().mul(100).round(1)

print("=== Monthly Revenue Trend ===")
print(monthly_trend.to_string())

print(f"\n📊 Best month : {monthly_trend['total_revenue'].idxmax().strftime('%B %Y')} "
      f"(${monthly_trend['total_revenue'].max():,.0f})")
print(f"📉 Worst month: {monthly_trend['total_revenue'].idxmin().strftime('%B %Y')} "
      f"(${monthly_trend['total_revenue'].min():,.0f})")
```

```python
# ════════════════════════════════════════════════════
# 🧹 Project 2: Data Cleaning Pipeline
#    (Realistic messy data → clean data)
# ════════════════════════════════════════════════════

# Messy raw data
raw = pd.DataFrame({
    'ID'     : ['001', '002', '003', '004', '005', '006', '007'],
    'Name'   : ['  alice ', 'BOB', 'charlie', None, 'EVE', 'frank', 'GRACE'],
    'Email'  : ['alice@gmail.com', 'bob@yahoo.com', 'INVALID', None,
                 'eve@gmail.com', 'frank@gmail.com', 'grace@yahoo.com'],
    'Age'    : ['25', '30', 'thirty-five', '28', '22', '40', '33'],
    'Salary' : ['55000', '72,000', '$85000', '68000', 'N/A', '95000', '61000'],
    'Dept'   : ['hr', 'IT', 'it ', 'Finance', 'HR', 'finance', 'IT'],
    'Joined' : ['2022-03-01', '2020/07/15', '01-10-2019', '11-20-2021', None,
                 '2018-09-30', '2021-04-15']
})

print("=== RAW (messy) data ===")
print(raw)
```

```python
# ── Step-by-step cleaning ──
clean = raw.copy()

# Step 1: Standardize Name
clean['Name'] = clean['Name'].str.strip().str.title()  # Strip + Title case

# Step 2: Validate & clean Email
valid_email = clean['Email'].str.contains(r'^[\w.]+@[\w.]+\.\w+$',
                                           na=False, regex=True)
clean['Email'] = clean['Email'].where(valid_email)  # Set invalid to NaN

# Step 3: Clean Age — coerce bad values to NaN
clean['Age'] = pd.to_numeric(clean['Age'], errors='coerce')

# Step 4: Clean Salary — remove $, commas, then convert
clean['Salary'] = (
    clean['Salary']
    .str.replace(r'[\$,]', '', regex=True)   # Remove $ and commas
    .pipe(pd.to_numeric, errors='coerce')     # Convert; bad values → NaN
)

# Step 5: Standardize Dept
dept_map = {'hr':'HR','it':'IT','finance':'Finance'}
clean['Dept'] = clean['Dept'].str.strip().str.lower().map(dept_map)

# Step 6: Parse dates with mixed formats
clean['Joined'] = pd.to_datetime(clean['Joined'],
                                  infer_datetime_format=True,
                                  errors='coerce')

print("=== AFTER CLEANING ===")
print(clean)
print("\n=== Remaining missing values ===")
print(clean.isna().sum())
```

### 💡 Key Takeaways
- **`.loc` is label-based and inclusive** on both ends; **`.iloc` is position-based and exclusive** on stop
- Always use `&`, `|`, `~` for boolean operations — never `and`, `or`, `not`
- **`groupby().agg()`** collapses rows; **`groupby().transform()`** keeps original row count
- Use **`pd.merge(how='left')`** as your default join — it keeps all left-side rows
- **`pd.to_numeric(errors='coerce')`** is the safest way to clean dirty numeric data
- **`pivot_table`** = long → wide; **`melt`** = wide → long — learn both!
- Always call **`.reset_index(drop=True)`** after filtering to avoid stale index labels
- Use **`category` dtype** for low-cardinality string columns — saves significant memory
